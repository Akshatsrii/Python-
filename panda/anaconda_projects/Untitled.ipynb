```python
import pandas as pd

data = {
    "Name": ["Akshat", "Rahul", "Priya", "Aman", "Neha"],
    "Age": [21, 22, 20, 23, 19],
    "Marks": [90, 75, 95, 65, 88],
    "City": ["Kota", "Delhi", "Jaipur", "Delhi", "Kota"]
}

df = pd.DataFrame(data)

print(df)
```

```python
df[df["Age"] > 20]
```

```python
df[df["Marks"] >= 90]
```

```python
df[(df["Age"] > 20) & (df["Marks"] > 80)]
```

```python
df[(df["City"] == "Delhi") | (df["Marks"] > 90)]
```

```python
df[~(df["City"] == "Delhi")]
```

```python
df[df["Name"].str.startswith("A")]
```

```python
df[df["Marks"].isnull()]
```

```python
df[(df["Age"] > 20) & (df["City"] == "Kota")]
```
