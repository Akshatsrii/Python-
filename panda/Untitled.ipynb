```python
# Concatenation (pd.concat())
# Used to join DataFrames vertically or horizontally.
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Akshat", "Rahul"],
    "Age": [21, 22]
})

df2 = pd.DataFrame({
    "Name": ["Priya", "Rohan"],
    "Age": [20, 23]
})

result = pd.concat([df1, df2])

print(result)
```

```python
df1 = pd.DataFrame({
    "Name": ["Akshat", "Rahul"]
})

df2 = pd.DataFrame({
    "Age": [21, 22]
})

result = pd.concat([df1, df2], axis=1)

print(result)
```

```python
df1 = pd.DataFrame({
    "Name": ["Akshat", "Rahul"]
})

df2 = pd.DataFrame({
    "Age": [21, 22]
})

pd.concat([df1, df2])
```

```python
# Merge (pd.merge())
students = pd.DataFrame({
    "ID":[1,2,3],
    "Name":["Akshat","Rahul","Priya"]
})
```

```python
marks = pd.DataFrame({
    "ID":[1,2,4],
    "Marks":[95,88,75]
})
```

```python
# Inner Merge
pd.merge(students, marks, on="ID")
```

```python
# Left Merge
# All rows from left dataframe.
pd.merge(students, marks, on="ID", how="left")
```

```python
# Right
pd.merge(students, marks, on="ID", how="right")
```

```python
# outer
pd.merge(students, marks, on="ID", how="outer")
```

```python
# . Merge on Different Column Names
df1 = pd.DataFrame({
    "Student_ID":[1,2],
    "Name":["Akshat","Rahul"]
})

df2 = pd.DataFrame({
    "ID":[1,2],
    "Marks":[90,80]
})

pd.merge(df1, df2,
         left_on="Student_ID",
         right_on="ID")
```

```python
# Join (join())

# Works using index.

df1 = pd.DataFrame({
    "Name":["Akshat","Rahul"]
}, index=[1,2])

df2 = pd.DataFrame({
    "Marks":[95,88]
}, index=[1,2])

df1.join(df2)
```

```python
# Missing Data

# # Missing values are represented by
df = pd.DataFrame({
    "Name":["Akshat","Rahul",None],
    "Age":[21,None,20]
})
```

```python
# Detect Missing Values
# isnull()
df.isnull()
```

```python
# notnull()
df.notnull()
```

```python
# Fill Missing Values
# Fill with Zero
df.fillna(0)
```

```python
# Fill with String
df.fillna("Unknown")
```

```python
# Fill Particular Column
df["Age"] = df["Age"].fillna(18)
```
