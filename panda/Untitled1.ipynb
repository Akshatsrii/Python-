```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Akshat", "Rahul", "Priya", "Rohan", "Neha"],
    "Department": ["IT", "IT", "HR", "HR", "IT"],
    "Salary": [50000, 60000, 45000, 40000, 55000]
})

print(df)
```

```python
pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    aggfunc="mean"
)

print(pivot)
```

```python
pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    aggfunc="sum"
)

print(pivot)
```

```python
pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    aggfunc="count"
)

print(pivot)
```

```python
df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "IT"],
    "Gender": ["M", "F", "F", "M", "M"],
    "Salary": [50000, 60000, 45000, 40000, 55000]
})

pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    columns="Gender",
    aggfunc="mean"
)

print(pivot)
```

```python
pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    aggfunc=["mean", "sum", "max", "min"]
)

print(pivot)
```

```python
df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR"],
    "Salary": [50000, 60000, 45000, 40000],
    "Bonus": [5000, 6000, 3000, 3500]
})

pivot = pd.pivot_table(
    df,
    values=["Salary", "Bonus"],
    index="Department",
    aggfunc="sum"
)

print(pivot)
```

```python
pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    columns="Gender",
    aggfunc="mean",
    fill_value=0
)
```

```python
pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    aggfunc="sum",
    margins=True
)

print(pivot)
```

```python
# Arithmetic Operations
# Example DataFrame
import pandas as pd

df = pd.DataFrame({
    "A": [10, 20, 30],
    "B": [5, 10, 15]
})

print(df)
```

```python
df["A"] * df["B"]
```

```python
df["A"] / df["B"]
```

```python
# Comparison Operations
df["A"] > 15
```

```python
df["A"] == 20
```

```python
# Logical Operations
# AND (&)
df[(df["A"] > 10) & (df["B"] > 5)]
```

```python
# OR (|)
df[(df["A"] > 20) | (df["B"] == 5)]
```

```python
# NOT (~)
df[~(df["A"] > 20)]
```

```python
# String Operations
df = pd.DataFrame({
    "Name": ["akshat", "rahul", "priya"]
})
# Upper Case
df["Name"].str.upper()
```

```python
# Lower Case
df["Name"].str.lower()
# Title Case
df["Name"].str.title()
```

```python
# Contains
df["Name"].str.contains("a")
# Replace
df["Name"].str.replace("a", "@")
```

```python
# Apply Operation
df = pd.DataFrame({
    "Marks": [70, 80, 90]
})

df["Marks"] = df["Marks"].apply(lambda x: x + 5)
```

```python
# Map Operation
df = pd.DataFrame({
    "Gender": ["M", "F", "M"]
})

mapping = {
    "M": "Male",
    "F": "Female"
}

df["Gender"] = df["Gender"].map(mapping)
```

```python
# 9. Sorting
# Ascending
df.sort_values("Marks")
```

```python
# Value Counts
df["Gender"].value_counts()
```

```python
# Unique Values
df["Gender"].unique()
```

```python
# Number of Unique Values
df["Gender"].nunique()
```

```python
# GroupBy Operation
df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR"],
    "Salary": [50000, 60000, 40000, 45000]
})

df.groupby("Department")["Salary"].sum()
```

```python
# Cumulative Operations
# Cumulative Sum
df["Salary"].cumsum()
```

```python
# 18. Correlation
df.corr(numeric_only=True)
```

```python
# Finding Data in Pandas (Selection & Filtering)

# Finding data means selecting specific rows, columns, or values from a DataFrame.

# Sample DataFrame
import pandas as pd

df = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Akshat", "Rahul", "Priya", "Rohan"],
    "Age": [21, 22, 20, 23],
    "Marks": [90, 85, 95, 80]
})

print(df)
```

```python
# Find a Column
df["Name"]
```

```python
# Find a Row using loc[]
# loc[] uses row labels (index names).
df.loc[2]
```

```python
# Find a Row using iloc[]

# iloc[] uses row positions.

df.iloc[1]
```

```python
# Find a Specific Value
df.loc[0, "Name"]
```

```python
# Find Rows by Condition

# Age greater than 21

df[df["Age"] > 21]
```

```python
# Find Data Using String Functions

# Names containing "a"

df[df["Name"].str.contains("a", case=False)]
```

```python
# 19. Find Largest Values

# Top 2 highest marks

df.nlargest(2, "Marks")
```

```python
# Find Smallest Values
df.nsmallest(2, "Marks")
```
