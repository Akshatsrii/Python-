```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as pit
```

```python
x= np.linspace(0,5,11)
x
y= x**2
y
```

```python
pit.plot(x,y)
```

```python
pit.plot (x,y)
pit.title("Hello")
pit.xlabel("Chacha")
pit.ylabel ("Chachi")
```

```python
pit.subplot(2,2,1)
pit.plot(x,y)
pit.subplot(2,2,4)
pit.plot(x**2, x**2)
pit.subplot(2,2,3)
pit.plot(y,x)
pit.subplot(2,2,2)
pit.plot(x,y*2)
```

```python
# objext oriented way 
fig, ax = pit.subplots()

# Plot data
ax.plot([1, 2, 3, 4], [10, 20, 25, 30])

# Add labels and title
ax.set_title("Simple Line Graph")
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")

# Display
pit.show()
```
