```python
pit.subplot(2,2,1)
pit.plot(x,y)
pit.subplot(2,2,4)
pit.plot(x**2, x**2)
pit.subplot(2,2,3)
pit.plot(y,x)
pit.subplot(2,2,2)
pit.plot(x,y*2)
```

```python
# objext oriented way 
fig, ax = pit.subplots()

# Plot data
ax.plot([1, 2, 3, 4], [10, 20, 25, 30])

# Add labels and title
ax.set_title("Simple Line Graph")
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")

# Display
pit.show()
```

```python
import seaborn as sns
df = sns.load_dataset('tips')

```

```python
df['size'].unique()
```

```python
sns.displot(data=df, x='total_bill', bins=10)
```

```python
sns.jointplot(data=df, x='total_bill', y='tip')
```

```python
sns.jointplot(data=df, x='total_bill', y='tip', kind='scatter')
# scatter
```

```python
# hexabin
sns.jointplot(data=df, x='total_bill', y='tip', kind='hex')
```

```python

# kde
sns.jointplot(data=df, x='total_bill', y='tip', kind='kde')
```

```python
# reggression
sns.jointplot(data=df, x='total_bill', y='tip', kind='reg')
```

```python
sns.pairplot(df)
```

```python
sns.pairplot(df, hue ="sex")
```

```python
# rugplot
sns.rugplot(df)
```

```python
sns.countplot(x=df['sex'],hue = df['smoker'])
# countplot
```

```python
sns.barplot(data=df, x='sex', y='total_bill')
# barplot
```

```python
# boxplot
sns.boxplot( x='tip', y='day', data= df)
```

```python
# volinplot
sns.violinplot(data=df, x='sex', y='total_bill')
```

```python
# stripplot
sns.stripplot(data=df, x='sex', y='total_bill')
```

```python
# swarm plot
sns.swarmplot(data=df, x='sex', y='total_bill')
```

```python
# heatmap
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=True)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.corr(numeric_only=True)

sns.clustermap(corr, annot=True)

plt.show()
```
